```python
import pandas as pd

# Memuat dataset
df = pd.read_csv('wbc.csv')

# Membuang kolom yang tidak dapat digunakan
df = df.drop(columns=['id', 'Unnamed: 32'])
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['diagnosis'] = le.fit_transform(df['diagnosis'])
```

```python
from sklearn.preprocessing import StandardScaler

# Memisahkan fitur (X) yang ingin disandardisasi (selain kolom diagnosis)
features = df.drop(columns=['diagnosis'])

# Melakukan standardisasi
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)

# Mengembalikan hasilnya ke dalam bentuk DataFrame Pandas
df_scaled = pd.DataFrame(scaled_features, columns=features.columns)

# Menggabungkan kembali dengan kolom target 'diagnosis'
df_final = pd.concat([df['diagnosis'].reset_index(drop=True), df_scaled], axis=1)

# Menampilkan beberapa baris data hasil akhir
df_final.head()
```
