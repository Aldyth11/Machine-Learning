PyPREP: Digunakan untuk melakukan preprocessing (prapemrosesan) data EEG (Electroencephalogram), seperti mendeteksi saluran yang rusak (noisy channels), mereferensi ulang data, dan membersihkan artefak.

Scipy: Merupakan library inti untuk komputasi saintifik dan matematika tingkat lanjut di Python yang sering dipakai untuk pemrosesan sinyal (seperti filter digital, transformasi Fourier), optimasi, dan perhitungan statistik.

wandb (Weights & Biases): Digunakan untuk melakukan experiment tracking, memvisualisasikan metrik pelatihan model machine learning atau deep learning, serta mencatat hyperparameter secara real-time.

pyECG: Digunakan untuk membaca, menulis, dan mengelola format file dataset standar rekam medis EKG/ECG (Electrocardiogram), seperti format Physiobank WFDB.
