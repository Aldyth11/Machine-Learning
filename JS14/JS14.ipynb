# Praktikum 1: Klasifikasi Cat vs. Dog

### Langkah 0: Unduh dan Ekstrak Dataset

```python
import zipfile
import os

with zipfile.ZipFile('dataset.zip', 'r') as zip_ref:
    zip_ref.extractall('.')
```

### Langkah 1: Import Library

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

### Langkah 2.1: Pra Pengolahan Data Training

```python
train_datagen = ImageDataGenerator(rescale = 1./255,
                                   shear_range = 0.2,
                                   zoom_range = 0.2,
                                   horizontal_flip = True)
training_set = train_datagen.flow_from_directory('dataset/training_set',
                                                 target_size = (64, 64),
                                                 batch_size = 32,
                                                 class_mode = 'binary')
```

### Langkah 2.2: Pra Pengolahan Data Testing

```python
test_datagen = ImageDataGenerator(rescale = 1./255)
test_set = test_datagen.flow_from_directory('dataset/test_set',
                                            target_size = (64, 64),
                                            batch_size = 32,
                                            class_mode = 'binary')
```

### Langkah 3.1: Inisiasi Model CNN

```python
cnn = tf.keras.models.Sequential()
```

### Langkah 3.2: Pembuatan Layer Konvolusi 1

```python
cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu', input_shape=[64, 64, 3]))
```

### Langkah 3.3: Pembuatan Layer Pooling 1

```python
cnn.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))
```

### Langkah 3.4: Pembuatan Layer Konvolusi 2 dan Pooling 2

```python
cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu'))
cnn.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))
```

### Langkah 3.5: Flattening

```python
cnn.add(tf.keras.layers.Flatten())
```

### Langkah 3.6: Fully Connected Layer 1 (Input)

```python
cnn.add(tf.keras.layers.Dense(units=128, activation='relu'))
```

### Langkah 3.7: Fully Connected Layer 2 (Output)

```python
cnn.add(tf.keras.layers.Dense(units=1, activation='sigmoid'))
```

### Langkah 3.8: Compile Model CNN

```python
cnn.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

### Langkah 4: Fit CNN

```python
cnn.fit(x = training_set, validation_data = test_set, epochs = 25)
```

### Langkah 5: Prediksi dengan 1 Citra

```python
import numpy as np
import pandas as pd
from tensorflow.keras.preprocessing import image

test_image = image.load_img('dataset/single_prediction/cat_or_dog_1.jpg', target_size = (64, 64))
test_image = image.img_to_array(test_image)
test_image = np.expand_dims(test_image, axis = 0)
result = cnn.predict(test_image)

if result[0][0] == 1:
    prediction = 'dog'
else:
    prediction = 'cat'

df_res = pd.DataFrame([{
    'Probability Output': result[0][0],
    'Predicted Class': prediction
}])
display(df_res)
```

# Tugas Praktikum 1: Modifikasi Model CNN (Akurasi Testing > 80%)

### Langkah 0: Ekstrak Dataset

```python
import gdown
import zipfile
import os

url = 'https://drive.google.com/uc?id=1vYrqBI1VmiXXJd5sgtKK2nuQvC8T1ryb'
gdown.download(url, 'dataset.zip', quiet=False)

with zipfile.ZipFile('dataset.zip', 'r') as zip_ref:
    zip_ref.extractall('.')
```

### Langkah 1: Import Library

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models
import pandas as pd
```

### Langkah 2.1: Pra Pengolahan Data Training dengan Augmentasi

```python
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=15,
    width_shift_range=0.1,
    height_shift_range=0.1,
    zoom_range=0.1,
    horizontal_flip=True
)

training_set = train_datagen.flow_from_directory(
    'dataset/training_set',
    target_size=(96, 96),
    batch_size=64,
    class_mode='binary'
)
```

### Langkah 2.2: Pra Pengolahan Data Testing

```python
test_datagen = ImageDataGenerator(rescale=1./255)

test_set = test_datagen.flow_from_directory(
    'dataset/test_set',
    target_size=(96, 96),
    batch_size=64,
    class_mode='binary'
)
```

### Langkah 3: Pembuatan Arsitektur Model CNN Tingkat Lanjut

```python
cnn = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(96, 96, 3)),
    layers.BatchNormalization(),
    layers.MaxPooling2D(2, 2),

    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D(2, 2),

    layers.Conv2D(128, (3, 3), activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D(2, 2),

    layers.Flatten(),
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(1, activation='sigmoid')
])
```

### Langkah 4: Compile Model CNN

```python
cnn.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

### Langkah 5: Pelatihan Model (Fit CNN)

```python
history = cnn.fit(
    x=training_set,
    validation_data=test_set,
    epochs=15
)
```

### Langkah 6: Evaluasi Performa Model

```python
test_loss, test_acc = cnn.evaluate(test_set)
df_eval = pd.DataFrame([{
    'Test Loss': test_loss,
    'Test Accuracy': test_acc
}])
display(df_eval)
```

# Tugas Praktikum 2: Klasifikasi Dataset MNIST Menggunakan CNN

### Langkah 1: Import Library dan Load Dataset MNIST

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import pandas as pd

mnist = tf.keras.datasets.mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

### Langkah 2: Pra Pengolahan Data MNIST

```python
X_train = X_train.reshape((X_train.shape[0], 28, 28, 1)).astype('float32') / 255.0
X_test = X_test.reshape((X_test.shape[0], 28, 28, 1)).astype('float32') / 255.0
```

### Langkah 3: Pembuatan Arsitektur Model CNN MNIST

```python
model_mnist = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])
```

### Langkah 4: Compile Model MNIST

```python
model_mnist.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

### Langkah 5: Pelatihan Model MNIST

```python
model_mnist.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_data=(X_test, y_test)
)
```

### Langkah 6: Evaluasi dan Prediksi Model MNIST

```python
test_loss, test_acc = model_mnist.evaluate(X_test, y_test, verbose=0)
predictions = model_mnist.predict(X_test[:5])
pred_classes = predictions.argmax(axis=1)

df_result = pd.DataFrame({
    'Actual Label': y_test[:5],
    'Predicted Label': pred_classes
})

df_metrics = pd.DataFrame([{
    'Test Loss': test_loss,
    'Test Accuracy': test_acc
}])

display(df_metrics)
display(df_result)
```
